# Introduction to Image Processing with Python

This notebook demonstrates basic image processing operations:
1. Loading an image
2. Converting to grayscale
3. Displaying images

## Import Required Libraries

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Configure matplotlib to display images inline
%matplotlib inline
````

## Create a Sample Image

For this tutorial, we'll create a simple colored sample image using NumPy.

````python
# Create a sample RGB image (200x200 pixels)
# This creates a gradient image with different colors
height, width = 200, 200
sample_image = np.zeros((height, width, 3), dtype=np.uint8)

# Create a gradient pattern
for i in range(height):
    for j in range(width):
        sample_image[i, j] = [i % 256, j % 256, (i + j) % 256]

print(f"Image shape: {sample_image.shape}")
print(f"Image data type: {sample_image.dtype}")
````

## Display the Original Image

````python
# Convert BGR to RGB for proper display with matplotlib
# Note: OpenCV uses BGR format, but matplotlib expects RGB
image_rgb = cv2.cvtColor(sample_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(image_rgb)
plt.title('Original Color Image')
plt.axis('off')
plt.show()
````

## Convert to Grayscale

Grayscale conversion reduces the image from 3 color channels (RGB) to a single intensity channel.

````python
# Convert the image to grayscale
gray_image = cv2.cvtColor(sample_image, cv2.COLOR_BGR2GRAY)

print(f"Grayscale image shape: {gray_image.shape}")
print(f"Original image shape: {sample_image.shape}")
````

## Display the Grayscale Image

````python
plt.figure(figsize=(6, 6))
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
````

## Side-by-Side Comparison

````python
# Display both images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(image_rgb)
axes[0].set_title('Original Color Image')
axes[0].axis('off')

axes[1].imshow(gray_image, cmap='gray')
axes[1].set_title('Grayscale Image')
axes[1].axis('off')

plt.tight_layout()
plt.show()
````

## Working with Real Images

To load an actual image file from disk, you can use:

```python
# Load an image from file
image = cv2.imread('../data/your_image.jpg')

# Check if image was loaded successfully
if image is None:
    print("Error: Could not load image")
else:
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # Display
    plt.imshow(gray, cmap='gray')
    plt.show()
```

Place your images in the `data/` folder and update the path accordingly.
