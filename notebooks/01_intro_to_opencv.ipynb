**Important**: OpenCV loads images in BGR (Blue-Green-Red) format, not RGB!

```python
# First, let's create a simple test image
def create_test_image():
    """Create a colorful test image for learning."""
    # Create a 300x300 image with 3 color channels (BGR)
    img = np.zeros((300, 300, 3), dtype=np.uint8)
    
    # Add colored rectangles (remember: BGR format!)
    img[50:150, 50:150] = [255, 0, 0]    # Blue rectangle
    img[50:150, 150:250] = [0, 255, 0]   # Green rectangle  
    img[150:250, 50:150] = [0, 0, 255]   # Red rectangle
    img[150:250, 150:250] = [255, 255, 0] # Cyan rectangle
    
    # Add some circles
    cv2.circle(img, (75, 225), 25, (255, 255, 255), -1)  # White circle
    cv2.circle(img, (225, 225), 25, (0, 0, 0), -1)       # Black circle
    
    return img

# Create and save the test image
test_image = create_test_image()
cv2.imwrite('../data/sample.jpg', test_image)
print("✅ Test image created and saved as 'data/sample.jpg'")

# TODO: Now load the image using OpenCV's image reading function
# Check the documentation for the function that reads image files
image_path = '../data/sample.jpg'
image_bgr = None  # Replace with your code to load the image

# TODO: Check if image was loaded successfully
# Images that fail to load will be None - check for this condition
if False:  # Replace with proper condition check
    print(f"❌ Error: Could not load image from {image_path}")
    print("Make sure the data/ directory exists and the file was created.")
else:
    print(f"✅ Image loaded successfully!")
    print(f"   Image shape: {image_bgr.shape}")
    print(f"   Image dimensions: {image_bgr.shape[1]}x{image_bgr.shape[0]} pixels")
    print(f"   Number of color channels: {image_bgr.shape[2]}")
    print(f"   Data type: {image_bgr.dtype}")
```
