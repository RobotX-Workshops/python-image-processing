# Introduction to OpenCV

This notebook demonstrates a simple image processing workflow using OpenCV. We'll load an image, convert it to grayscale, and display the results side-by-side.

## Learning Objectives
- Import and use OpenCV (cv2) library
- Load and display images
- Convert images to grayscale
- Visualize results using Matplotlib

## Step 1: Import Required Libraries

We'll need the following libraries:
- **cv2**: OpenCV library for image processing
- **numpy**: For numerical operations on image arrays
- **matplotlib**: For displaying images

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Configure matplotlib for inline display
%matplotlib inline
```

## Step 2: Load the Image

We'll load a sample image from the `data/` directory. OpenCV's `imread()` function reads the image file and returns it as a NumPy array.

**Note**: OpenCV loads images in BGR (Blue-Green-Red) color space by default, not RGB.

```python
# Load the image
image_path = '../data/sample.jpg'
image_bgr = cv2.imread(image_path)

# Check if image was loaded successfully
if image_bgr is None:
    print(f"Error: Could not load image from {image_path}")
    print("Please make sure sample.jpg exists in the data/ directory.")
else:
    print(f"Image loaded successfully!")
    print(f"Image shape: {image_bgr.shape}")
    print(f"Image dimensions: {image_bgr.shape[1]}x{image_bgr.shape[0]} pixels")
```

## Step 3: Convert BGR to RGB

Since Matplotlib expects images in RGB format and OpenCV loads them in BGR, we need to convert the color space for proper display.

```python
# Convert BGR to RGB for matplotlib display
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
```

## Step 4: Convert to Grayscale

Converting an image to grayscale reduces it from 3 color channels (RGB) to a single channel, representing intensity values from 0 (black) to 255 (white).

```python
# Convert the image to grayscale
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)

print(f"Original image shape: {image_rgb.shape}")
print(f"Grayscale image shape: {image_gray.shape}")
```

## Step 5: Display Images Side-by-Side

We'll use Matplotlib to create a figure with two subplots showing the original and grayscale images side-by-side for comparison.

```python
# Create a figure with two subplots
plt.figure(figsize=(12, 5))

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title('Original Image')
plt.axis('off')

# Display grayscale image
plt.subplot(1, 2, 2)
plt.imshow(image_gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
